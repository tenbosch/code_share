# SQL AI Functions

This notebook demonstrates built-in Databricks SQL AI functions — classification, summarization, masking, sentiment analysis, and generation.

## Classification
Uses `ai_classify` to categorize each transcription into one of three labels: **billing**, **shipment**, or **other**. This is useful for routing customer calls or tickets to the appropriate team.

```python
%sql
select
  file_name,
  ai_classify(transcription, '["billing", "shipment", "other"]', MAP('version', '2.1'))
from
  tenbosch.aips.audio_transcriptions;
```

## Summarization
Uses `ai_summarize` to generate a concise summary of each transcription, limited to approximately 20 words. Helpful for quickly reviewing call content without reading the full text.

```python
%sql
select
  file_name,
  ai_summarize(transcription, 20)
from
  tenbosch.aips.audio_transcriptions;
```

## PII Masking
Uses `ai_mask` to redact person names from each transcription. This helps anonymize data for compliance or sharing purposes while preserving the rest of the content.

```python
%sql
select
  file_name,
  ai_mask(transcription, array('person'))
from
  tenbosch.aips.audio_transcriptions;
```

## Sentiment Analysis
Uses `ai_analyze_sentiment` to determine the overall sentiment of each **transcription** (e.g., positive, negative, neutral, mixed). Useful for identifying frustrated callers or measuring customer satisfaction.

```python
%sql
select
  file_name,
  ai_analyze_sentiment(transcription)
from
  tenbosch.aips.audio_transcriptions;
```

## Generate Next Steps
Uses `ai_gen` to generate recommended next steps based on the content of each **transcription**. This is useful for automating post-call action items (such as follow-up tasks, escalations, or resolutions) ensuring that no critical steps are missed after a customer interaction.

```python
%sql
select
  file_name,
  ai_gen(concat("What should the next steps be from this call: ", transcription))
from
  tenbosch.aips.audio_transcriptions;
```
